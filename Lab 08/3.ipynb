```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv")
df["scenario"] = df["scenario"].astype(str).str.strip()
data = df[df["scenario"] == "03_SalaryDiff"].copy()
print(data.shape)
data.head()
```

### Box plot

```python
data["prep_course"] = data["prep_course"].astype(str).str.strip().str.title()
data["exam_score"] = pd.to_numeric(data["exam_score"], errors="coerce")

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=data,
    x="prep_course",
    y="exam_score",
    order=["Entry", "Mid", "Senior"],
    hue="prep_course",
    hue_order=["Entry", "Mid", "Senior"],
    palette="Set2",
    legend=False,
)

plt.title("Salary Spread Across Experience Levels")
plt.xlabel("Experience Level")
plt.ylabel("Salary ($)")
plt.tight_layout()
plt.show()
```
