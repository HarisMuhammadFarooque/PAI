```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = pd.read_csv("data.csv")
```

```python
filtered_data = data[data['scenario'] == '01_StudyVsScore']
filtered_data
```

```python
sns.scatterplot(data= filtered_data, 
                 x='study_hours',
                 y='exam_score',
                 hue= 'prep_course',
                 markers={'Completed': 'o', 'None': 'X'},
                 style= 'prep_course')
plt.title('Study Hours vs. Exam Score by Prep Course Status')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
```
