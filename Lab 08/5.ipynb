# Lab 08 - Task 5: Ad Spend vs Revenue Regression Plot (Seaborn)

Measure advertising ROI.

Column mapping: `study_hours` = marketing budget (ad spend), `exam_score` = revenue.

**Note:** place `data.csv` in the same folder as this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv")
df["scenario"] = df["scenario"].astype(str).str.strip()
data = df[df["scenario"] == "05_AdVsRevenue"].copy()
print(data.shape)
data.head()
```

### Regression plot

```python
data["study_hours"] = pd.to_numeric(data["study_hours"], errors="coerce")
data["exam_score"] = pd.to_numeric(data["exam_score"], errors="coerce")

plt.figure(figsize=(8, 5))
sns.regplot(
    data=data,
    x="study_hours",
    y="exam_score",
    color="darkgreen",
    marker="+",
    line_kws={"color": "red"},
)

plt.title("Impact of Advertising Spend on Total Revenue")
plt.xlabel("Ad Spend ($)")
plt.ylabel("Revenue ($)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
