```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv")
df["scenario"] = df["scenario"].astype(str).str.strip()
data = df[df["scenario"] == "04_CustAgeDist"].copy()
print(data.shape)
data.head()
```

```python
data = df[df["scenario"].astype(str).str.strip() == "04_CustAgeDist"].copy()
data["prep_course"] = pd.to_numeric(data["prep_course"], errors="coerce")
data["study_hours"] = data["study_hours"].astype(str).str.strip()

plt.figure(figsize=(8, 5))
sns.histplot(
    data=data,
    x="prep_course",
    hue="study_hours",
    bins=5,
    kde=True,
)

plt.title("Customer Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Customer Count")
plt.tight_layout()
plt.show()
```
