**DataSet Generation**

```python
import pandas as pd
import numpy as np
 
np.random.seed(42)
n_rows = 150
data = {
    'Customer_ID': [f'CUST_{1000 + i}' if i % 12 != 0 else np.nan for i in range(n_rows)],
    'Join_Date': pd.date_range(start='2023-01-01', periods=n_rows, freq='2D'),
    'Age': [np.random.randint(18, 65) if i % 7 != 0 else np.nan for i in range(n_rows)],
    'Gender': np.random.choice(['Male', 'Female', 'Other'], size=n_rows, p=[0.48, 0.48, 0.04]),
    'City': np.random.choice(['Karachi', 'Lahore', 'Islamabad', 'Peshawar'], size=n_rows),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Home', 'Books'], size=n_rows),
    'Order_Count': np.random.randint(1, 30, size=n_rows),
    'Total_Spent': [round(np.random.uniform(500, 50000), 2) if i % 9 != 0 else np.nan for i in range(n_rows)],
    'Is_VIP': np.random.choice([True, False], size=n_rows, p=[0.2, 0.8])
}
df = pd.DataFrame(data)
df.head(10)
```

```python
print("Dataset Ready! Total rows:", len(df))
```

**ALL TASKS**


```python
print('Data Types: ', df.dtypes)
print('Shape: ', df.shape)
print('Data Types: ', df.isna().sum())
```

```python
df = df.dropna(subset='Customer_ID')
df
```

```python
val = df['Age'].median()
df['Age'] = df['Age'].fillna(val)

df['Total_Spent'] = df['Total_Spent'].fillna(df['Total_Spent'].mean())
df
```

```python
dt = pd.to_datetime(df['Join_Date'])
year = df['Join_Date'].dt.year
month = df['Join_Date'].dt.month

df['Join_Year'] = year
df['Join_Month'] = month
df
```

```python
df['Avg_Order_Value'] = df['Total_Spent']/df['Order_Count']
df
```

```python
df['High_Value_Customer'] = df['Avg_Order_Value'] > 2500
df
```

```python
gc = df.groupby('City')
print('------- Customer Count --------\n',gc['Customer_ID'].count())

avg = gc['Total_Spent'].mean()
print('------- Average Total Spent -------\n', avg)

maxCount = gc['Order_Count'].max()
print('------- Maximum Order Count -------\n', maxCount)
```

```python
pivot = pd.pivot_table(df, index='Category', columns='Is_VIP', values='Total_Spent', aggfunc='sum')
pivot
```

```python
top5 = (df[df['Category'].isin(['Electronics', 'Clothing'])]
        .sort_values('Total_Spent', ascending=False)
        .head(5))
top5[['Customer_ID', 'City', 'Category', 'Total_Spent']]
```

```python
vip = df[(df['Is_VIP']) & (df['City'].isin(['Karachi', 'Lahore']))]
vip.groupby('City')['Order_Count'].agg(['count', 'mean'])
```

```python
print("Avg spend by VIP:\n", df.groupby('Is_VIP')[['Total_Spent', 'Order_Count', 'Avg_Order_Value']].mean())
print("\nRevenue by city:\n", df.groupby('City')['Total_Spent'].sum().sort_values(ascending=False))

print("\nRevenue by category:\n", df.groupby('Category')['Total_Spent'].sum().sort_values(ascending=False))
print("\nHigh-value share: %.1f%%" % (df['High_Value_Customer'].mean() * 100))
```

**Insights**
- Lahore is the largest revenue source (≈ 1.06M, 41 customers), while Karachi has the highest average spend per customer (27,379).
- VIP and non-VIP customers spend almost the same on average (24,798 vs 24,839), and VIPs have a lower Avg_Order_Value (2,184) than non-VIPs (2,936).
- Books (≈ 1.00M) and Electronics (≈ 997K) lead in revenue, with Home last (≈ 614K); in every category, most of the revenue comes from non-VIP customers.
